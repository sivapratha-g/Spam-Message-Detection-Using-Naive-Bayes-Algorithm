```python

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
data={
    "message":[ "win a free prize now",
               "congratulations you won a prize",
               "Clain your free gift today",
               "Free entry into the contezt",
               "win money with this offer",
               "Claim your reward now",
               "Meeting is scheduled for tomorrow",
               "Please send me the report",
               "Are you coming to class today",
               "Your assignment is completed",
                "Let us meet for lunch",
                "Please call me when you arrive",
                "Free cash offer available",
                "you won a free reward",
                "claim your prize today",
                "The class starts at ten",
                "Please check you email",
                "I will send the file today",
                "Win a gift card now","we have a meeting tomorrow"

               ],
    "label":["spam",
             "spam",
             "spam",
             "spam",
             "spam",
             "spam",
             "not spam",
             "not spam",
             "not spam",
             "not spam","not spam",
             "not spam","spam","spam","spam",
             "not spam","not spam","not spam","spam","not spam"]
}
df=pd.DataFrame(data)
df.head()
```

```python
x=df["message"]
y=df["label"]
print(x.head())
print(y.head())
```

```python
vectorizer=CountVectorizer()
x_vectorized=vectorizer.fit_transform(x)
print(vectorizer.get_feature_names_out())
print(x_vectorized.toarray())
```

```python
x_train,x_test,y_train,y_test=train_test_split(

x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
vectorizer=CountVectorizer()
x_train_vec=vectorizer.fit_transform(x_train)
x_test_vec=vectorizer.transform(x_test)
model=MultinomialNB()
model.fit(x_train_vec,y_train)
```

```python
y_pred=model.predict(x_test_vec)
print("Actual labels:")
print(y_test.to_list())
print("predicted labels:")
print(y_pred)
```

```python
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy :",accuracy)
print(classification_report(y_test,y_pred,zero_division=0))
```
